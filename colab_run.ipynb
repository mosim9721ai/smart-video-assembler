# 🎬 Smart Video Assembler — Colab Runner

**Script → MP4 in one click.** Just paste your Fal.ai key + script, hit `Runtime → Run all`.

1. Get a free Fal.ai key at https://fal.ai/dashboard/keys (format `KEY_ID:KEY_SECRET`)
2. Paste it in cell 2
3. Paste your script in cell 3
4. `Runtime → Run all` (Ctrl+F9)
5. Wait 1-3 min → the MP4 appears + auto-downloads

```
# 1. Install ffmpeg + clone repo
!apt-get -qq install -y ffmpeg
!git clone -q --branch claude/google-ai-studio-pro-api-9arv38 https://github.com/mosim9721ai/smart-video-assembler.git repo
%cd repo
!pip install -q -r requirements.txt
```

```
# 2. Paste your Fal.ai API key here (hidden input)
import os, getpass
os.environ['FAL_KEY'] = getpass.getpass('Paste FAL_KEY (KEY_ID:KEY_SECRET): ').strip()
print('✅ FAL_KEY set' if os.environ.get('FAL_KEY') else '❌ empty')
```

```
# 3. Paste your script here (or leave the sample)
SCRIPT = """Take a breath right now.

A normal one.

Now imagine you stop.

No more breaths.

At first, you feel completely fine.

But here's the terrifying part:

Your body doesn't panic because you're running out of oxygen.

It panics because carbon dioxide is building up in your blood."""

# Options
STYLE = 'cinematic, dark moody, dramatic side-lighting, medical thriller aesthetic, film grain, photorealistic, 4k'
RATIO = '9:16'   # 9:16 for Shorts/Reels, 16:9 for YouTube, 1:1 for feed
FPS   = '30'

open('my_script.txt', 'w').write(SCRIPT)
print(f'✅ Script saved ({len(SCRIPT)} chars)')
```

```
# 4. Generate the video (this is the slow step — 1-5 min)
import os
os.environ['STYLE'] = STYLE
os.environ['RATIO'] = RATIO
os.environ['FPS']   = FPS

!python cli.py my_script.txt output.mp4
```

```
# 5. Preview the video inline + auto-download
from IPython.display import Video, display
from google.colab import files

display(Video('output.mp4', embed=True, width=360))
files.download('output.mp4')
files.download('output.srt')
```
