# 🎬 Smart Video Assembler — Colab Runner

## Two backends — pick one:

| Backend | Cost | Quality | Setup |
|---------|------|---------|-------|
| **`free`** (default) | 🆓 Zero | Very good (Flux via Pollinations + Edge-TTS) | Nothing |
| **`fal`** | ~$0.05/video | Best | Needs FAL_KEY with balance |

## Steps:

1. Upper right corner mein **"Connect"** button tap karo — wait for green tick.
2. `Runtime → Run all` (ya `Ctrl+F9`).
3. Cell 2 mein form aayega — Backend `free` chuno agar paisa nahi lagana.
4. Wait 2-5 min → video download.

```
# ============= CELL 1 =============
# Install ffmpeg + clone repo (1-2 min first time)
print('Step 1/3: installing ffmpeg…')
!apt-get -qq install -y ffmpeg > /dev/null 2>&1
print('   ✅ ffmpeg installed')

print('Step 2/3: cloning repo…')
!rm -rf repo && git clone -q --branch claude/google-ai-studio-pro-api-9arv38 https://github.com/mosim9721ai/smart-video-assembler.git repo
%cd repo
print('   ✅ repo cloned')

print('Step 3/3: installing Python packages…')
!pip install -q -r requirements.txt 2>&1 | tail -3
print('   ✅ ready!')
print('\n👉 Ab cell 2 chalao.')
```

```
# ============= CELL 2 — configuration =============
# Yaha form fields dikhengey. Sirf form pe tap karke change karo.

BACKEND = "free"   #@param ["free", "fal"]
FAL_KEY = ""       #@param {type:"string"}
STYLE   = "cinematic, dark moody, dramatic side-lighting, medical thriller aesthetic, film grain, photorealistic, 4k"   #@param {type:"string"}
RATIO   = "9:16"   #@param ["9:16", "16:9", "1:1"]
FPS     = "30"     #@param ["24", "30", "60"]

import os
os.environ['BACKEND'] = BACKEND
os.environ['STYLE']   = STYLE
os.environ['RATIO']   = RATIO
os.environ['FPS']     = FPS
if FAL_KEY.strip():
    os.environ['FAL_KEY'] = FAL_KEY.strip()

print(f'🔧 Backend: {BACKEND}')
if BACKEND == 'free':
    print('   Pollinations.ai (Flux) + Edge-TTS — no key, no billing.')
elif BACKEND == 'fal':
    if os.environ.get('FAL_KEY'):
        print(f'   ✅ FAL_KEY set ({len(os.environ["FAL_KEY"])} chars)')
    else:
        print('   ❌ BACKEND=fal ke liye FAL_KEY zaroori hai. Form field mein paste karo.')
print(f'✅ ratio = {RATIO}, fps = {FPS}')
print(f'✅ style = {STYLE[:80]}…')
```

```
# ============= CELL 3 — your script =============
# Apna script yaha paste karo (triple quotes ke andar).

SCRIPT = """Take a breath right now.

A normal one.

Now imagine you stop.

No more breaths.

At first, you feel completely fine.

But here's the terrifying part:

Your body doesn't panic because you're running out of oxygen.

It panics because carbon dioxide is building up in your blood.

Around 30 seconds in, your brain starts screaming: Breathe.

Your diaphragm begins contracting, trying to force you to inhale.

And while all of this is happening, your oxygen is quietly dropping.

That's the dangerous part.

Because carbon dioxide creates the warning, but low oxygen can make you unconscious.

And if you hyperventilate before holding your breath, you can silence that warning while your oxygen keeps falling.

That's how people can black out underwater without feeling the danger coming.

So don't test this in water.

Your body's urge to breathe isn't weakness.

It's your survival system trying to save you."""

with open('my_script.txt', 'w') as f:
    f.write(SCRIPT)
print(f'✅ script saved ({len(SCRIPT)} chars, {SCRIPT.count(chr(10))+1} lines)')
```

```
# ============= CELL 4 — generate video =============
# Slow step: 2-5 min. Progress dikhega neeche.

!python cli.py my_script.txt output.mp4
```

```
# ============= CELL 5 — preview + download =============
from IPython.display import Video, display
from google.colab import files
import os

if os.path.exists('output.mp4'):
    print(f'✅ Video ready! {os.path.getsize("output.mp4")/1024:.0f} KB')
    display(Video('output.mp4', embed=True, width=360))
    files.download('output.mp4')
    if os.path.exists('output.srt'):
        files.download('output.srt')
else:
    print('❌ output.mp4 nahi bana — cell 4 ka error dekho upar.')
```
